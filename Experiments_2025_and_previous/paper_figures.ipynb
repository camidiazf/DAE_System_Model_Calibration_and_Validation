```python
import pandas as pd
import os


# Folder where solutions are available
archivo_normal = os.path.join(os.getcwd(), "M_smegmatis_normal","M_smegmatis_normal_Results","FINAL_1","original_model","Solution_original.csv")
archivo_N_def = os.path.join(os.getcwd(), "M_smegmatis_N_def","M_smegmatis_N_def_Results","FINAL_1","original_model","Solution_original.csv")



# Read the CSV files into DataFrames
df_normal = pd.read_csv(archivo_normal)
df_N_def = pd.read_csv(archivo_N_def)

# Experimental data for normal conditions
exp_data_path = os.path.join(os.getcwd(), 'Experimental_data.xlsx')

df_val_normal = pd.read_excel(exp_data_path, sheet_name='V_Normal') # V or VAL = PARAMETER VALIDATION
df_val_N_def = pd.read_excel(exp_data_path, sheet_name='V_Amm_def_rep') # V or VAL = PARAMETER VALIDATION

df_val_N_def = df_val_N_def.rename(columns={df_val_N_def.columns[0]: "t"})
df_val_normal = df_val_normal.rename(columns={df_val_normal.columns[0]: "t"})
```

```python
import matplotlib.pyplot as plt

variables = ["X", "C", "N", "pH"]

sim_dfs = {
    "Normal": df_normal,
    "Ammonium starvation": df_N_def,
}

val_dfs = {
    "Normal": df_val_normal,
    "Ammonium starvation": df_val_N_def,
}

fig, axes = plt.subplots(
    nrows=4,
    ncols=2,
    figsize=(10, 13),
    sharex="col"
)

for col, condition in enumerate(["Normal", "Ammonium starvation"]):
    df_sim = sim_dfs[condition]
    df_val = val_dfs[condition]

    for row, var in enumerate(variables):
        ax = axes[row, col]

        # Simulation line
        ax.plot(
            df_sim["t"],
            df_sim[var],
            label="Simulation",
            linewidth=1,
            color="black",
        )

        # Experimental / validation points
        ax.scatter(
            df_val["t"],
            df_val[var],
            label="Validation",
            s=45,
            color="black",
            zorder=3
        )

        ax.set_ylabel(var + " (g/L)", fontsize=15)
        ax.grid(True, alpha=0.3)

        if row == 0:
            ax.set_title(condition, fontsize=17)

        if row == len(variables) - 1:
            ax.set_xlabel("Time (h)", fontsize=15)

        if row == 0 and col == 0:
            ax.legend(loc="lower right")

fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Variables
variables = ["X", "C", "N", "pH", "Global"]

# NRMSE values
diaz_normal = [0.0952, 0.0951, 0.1960, 0.0473, 0.0197]
reformulated_normal = [0.0845, 0.0824, 0.2036, 0.1069, 0.0181]
diaz_amm = [0.4251, 0.3805, 0.0677, 0.2662, 0.0694]
reformulated_amm = [0.1498, 0.0916, 0.0418, 0.1358, 0.0187]

x = np.arange(len(variables))
width = 0.19

fig, ax = plt.subplots(figsize=(6, 6))

# Requested order:
# Diaz-normal, Reformulated-normal, Diaz-amm, Reformulated-amm

bars1 = ax.bar(
    x - 1.5 * width,
    diaz_normal,
    width,
    label="Díaz model – Normal",
    color="0.15",
    edgecolor="black"
)

bars2 = ax.bar(
    x - 0.5 * width,
    reformulated_normal,
    width,
    label="Reformulated model – Normal",
    color="0.55",
    edgecolor="black",
    hatch="..."
)

bars3 = ax.bar(
    x + 0.5 * width,
    diaz_amm,
    width,
    label="Díaz model – Amm. deficiency",
    color="0.40",
    edgecolor="black",
    hatch="///"
)

bars4 = ax.bar(
    x + 1.5 * width,
    reformulated_amm,
    width,
    label="Reformulated model – Amm. deficiency",
    color="0.85",
    edgecolor="black",
    hatch="\\\\\\"
)


ax.set_xlabel("Variable", fontsize=12)
ax.set_ylabel("NRMSE", fontsize=12)
ax.set_title(
    "Variable and Global NRMSE Across Models and Conditions",
    fontsize=14
)

ax.set_xticks(x)
ax.set_xticklabels(variables, fontsize=11)

ax.set_ylim(0, 0.49)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.35
)

ax.legend(
    frameon=False,
    ncol=1,
    fontsize=9,
    loc="upper right"
)

plt.tight_layout()
plt.show()
```
